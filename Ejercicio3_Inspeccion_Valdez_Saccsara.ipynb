## Ejercicio 3: Inspección visual de un producto

### Ficha PEAS
**Percepción (S)** = Matriz de 5x5 píxeles (0 = normal, 1 = defectuoso) capturada por una cámara 
**Acciones (A)** = Aprobar el producto, Enviar a revisión manual, Rechazar el producto 
**Entorno (E)** = Línea de producción industrial, determinista, estático, episódico 
**Objetivo** = Detectar productos defectuosos antes de que salgan al mercado, minimizando falsos aprobados 
**Medida de desempeño (P)** = Porcentaje de defectos detectados correctamente, cantidad de productos buenos enviados a revisión innecesaria, tiempo de inspección 


### Justificación de las reglas

Se usa el número de píxeles defectuosos como criterio porque es una medida objetiva y fácil de calcular a partir de la imagen capturada, un producto sin ningún defecto se aprueba directamente porque no representa riesgo. Entre 1 y 3 píxeles defectuosos se envía a revisión manual porque podría tratarse de un defecto menor o un error de la cámara, y conviene que una persona lo confirme antes de descartar el producto. 
Más de 3 píxeles defectuosos se rechaza automáticamente porque ya representa un daño significativo y visible, que no justifica el costo de una revisión manual.
