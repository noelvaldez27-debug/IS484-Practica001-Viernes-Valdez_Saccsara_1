## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS
**Percepción (S)** = Tiempo de espera en minutos, nivel de urgencia (baja,media y alta), si el cliente es premium
**Acciones (A)** = Escalar a nivel 1, Escalar a nivel 2 y Escalar a nivel 3
**Entorno (E)** = Sistema de mesa de ayuda; accesible, determinista, estático, episódico 
**Objetivo** = Priorizar tickets urgentes y clientes premium sin saturar los niveles superiores de soporte 
**Medida de desempeño (P)** = Tiempo promedio de resolución, satisfacción del cliente, correcta priorización de casos urgentes 



### Justificación de las reglas

La urgencia alta se escala de inmediato al nivel máximo porque representa el mayor riesgo para el negocio o el cliente, sin importar el tiempo de espera. La urgencia media depende del tiempo de espera porque un ticket que lleva esperando mucho tiempo merece mayor atención aunque no sea crítico. El estatus premium actúa como un acelerador adicional: un cliente premium con espera prolongada se escala antes que uno regular en la misma situación, reflejando el valor comercial de retener a esos clientes sin descuidar completamente al resto.

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    elif nivel_urgencia == "media":
        if cliente_premium:
            if tiempo_espera_minutos > 15:
                return "escalar a nivel 3", "urgencia media, cliente premium con espera prolongada"
            else:
                return "escalar a nivel 2", "urgencia media, cliente premium con espera corta"
        else:
            if tiempo_espera_minutos > 30:
                return "escalar a nivel 2", "urgencia media, espera prolongada"
            else:
                return "escalar a nivel 1", "urgencia media, espera corta"

    else:  # urgencia baja
        if cliente_premium:
            return "escalar a nivel 2", "urgencia baja pero cliente premium"
        else:
            return "escalar a nivel 1", "urgencia baja, cliente regular"
```
